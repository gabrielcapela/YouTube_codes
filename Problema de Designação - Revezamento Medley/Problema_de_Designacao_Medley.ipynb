#Problema de Designação - Revezamento Medley

```python
import pulp as pl
```

## Dados do Problema

```python
nadadores = ["Antônio", "Bruno", "Carlos", "Daniel", "Eduardo"]

estilos = ["COSTAS", "PEITO", "BORBOLETA", "LIVRE"]

tempos = {
    ("Antônio", "COSTAS"): 50.35,
    ("Antônio", "PEITO"): 56.79,
    ("Antônio", "BORBOLETA"): 48.70,
    ("Antônio", "LIVRE"): 45.85,

    ("Bruno", "COSTAS"): 50.95,
    ("Bruno", "PEITO"): 58.30,
    ("Bruno", "BORBOLETA"): 51.28,
    ("Bruno", "LIVRE"): 47.35,

    ("Carlos", "COSTAS"): 52.16,
    ("Carlos", "PEITO"): 59.25,
    ("Carlos", "BORBOLETA"): 50.30,
    ("Carlos", "LIVRE"): 46.10,

    ("Daniel", "COSTAS"): 51.28,
    ("Daniel", "PEITO"): 57.90,
    ("Daniel", "BORBOLETA"): 49.50,
    ("Daniel", "LIVRE"): 46.79,

    ("Eduardo", "COSTAS"): 49.92,
    ("Eduardo", "PEITO"): 56.40,
    ("Eduardo", "BORBOLETA"): 49.95,
    ("Eduardo", "LIVRE"): 45.90,
}
```

## Criação do modelo

```python
modelo = pl.LpProblem(
    "Problema_de_Designacao_Medley",
    pl.LpMinimize
)
```

### Variáveis de decisão

```python

# x[i,j] = 1 se o nadador i
# for designado ao estilo j

x = pl.LpVariable.dicts(
    "x",
    [(i, j) for i in nadadores for j in estilos],
    cat="Binary"
)
```

### Função objetivo


```python
# Minimizar o tempo total
modelo += pl.lpSum(
    tempos[(i, j)] * x[(i, j)]
    for i in nadadores
    for j in estilos
)
```

### Restrições

```python
# Restrição 1:
# Cada nadador nada no máximo em 1 estilo


for i in nadadores:
    modelo += pl.lpSum(
        x[(i, j)] for j in estilos
    ) <= 1, f"Nadador_{i}_um_estilo"

# Restrição 2:
# Cada estilo recebe exatamente 1 nadador

for j in estilos:
    modelo += pl.lpSum(
        x[(i, j)] for i in nadadores
    ) == 1, f"Um_nadador_no_estilo_{j}"
```

## Resolvendo o Problema

```python
modelo.solve()
```

### Imprimindo os resultados

```python
print("Status:", pl.LpStatus[modelo.status])
print("\nDesignação ótima:\n")

tempo_total = 0

for j in estilos:
    for i in nadadores:
        if x[(i, j)].value() == 1:
            tempo = tempos[(i, j)]
            tempo_total += tempo

            print(f"{i:10s} --> {j:12s} | Tempo = {tempo:.2f} s")

print(f"\nTempo total da equipe: {tempo_total:.2f} s")
```
